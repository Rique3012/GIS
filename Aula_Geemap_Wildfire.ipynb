#### Coordenadas retiradas do WHP 2020 reorientado no EPSG 4326

```python
aoi = ee.Geometry.Rectangle([[-127.9722015931231454,22.7665837474344066], [-65.2603715931231534,51.6327937474344054]])
```

```python
gridmet = ee.ImageCollection('IDAHO_EPSCOR/GRIDMET') \
.filter(ee.Filter.date('2020-01-01', '2021-01-01')) \
.filterBounds(aoi) \
.mean()

maximumTemperature = gridmet.select('tmmx')
maximumTemperatureVis = {
    "min": 233.08,
    "max": 327.14,
    "palette": ['d8d8d8', '4addff', '5affa3', 'f2ff89', 'ff725c'],
}

minimumTemperature = gridmet.select('tmmn')
maximumTemperatureVis = {
    "min": 225.54,
    "max": 314.88,
    "palette": ['d8d8d8', '4addff', '5affa3', 'f2ff89', 'ff725c'],
}

windSpeed = gridmet.select('vs')
windSpeedVis = {
    "min": 0.14,
    "max": 29.13,
    "palette": ['d8d8d8', '4addff', '5affa3', 'f2ff89', 'ff725c'],
}

precipitationAmount = gridmet.select('pr')
precipitationAmountVis = {
    "min": 0,
    "max": 690.44,
    "palette": ['d8d8d8', '4addff', '5affa3', 'f2ff89', 'ff725c'],
}

specificHumidity = gridmet.select('sph')
specificHumidityVis = {
    "min": 0,
    "max": 0.02,
    "palette": ['d8d8d8', '4addff', '5affa3', 'f2ff89', 'ff725c'],
}

Map.setCenter(-115.356, 38.686, 5)
Map.addLayer(maximumTemperature, maximumTemperatureVis, 'Maximum Temperature (K)')
Map.addLayer(minimumTemperature, maximumTemperatureVis, 'Minimum Temperature (K)')
Map.addLayer(windSpeed, windSpeedVis, 'Wind Speed (m/s)')
Map.addLayer(precipitationAmount, precipitationAmountVis, 'Precipitation Amount (mm)')
Map.addLayer(specificHumidity, specificHumidityVis, 'Specific Humidity (mass fraction)')
```

```python
def maskS2clouds(image):
    qa = image.select('QA60')

    # Bits 10 and 11 are clouds and cirrus, respectively.
    cloudBitMask = 1 << 10
    cirrusBitMask = 1 << 11

    # Both flags should be set to zero, indicating clear conditions.
    mask = qa.bitwiseAnd(cloudBitMask).eq(0) \
    .And(qa.bitwiseAnd(cirrusBitMask).eq(0))

    return image.updateMask(mask).divide(10000)

sentinel = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
.filterDate('2020-01-01', '2021-01-01') \
.filterBounds(aoi) \
.filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE',20)) \
.map(maskS2clouds) \
.mean()

bands = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12']
for band in bands:
  visualization = {
    "min": 0.0,
    "max": 0.3,
    "bands": band,
  }
  Map.addLayer(sentinel, visualization, band)
```

#### Cálculo de índices

```python
ndvi = sentinel.normalizedDifference(['B8', 'B4'])
nbr = sentinel.normalizedDifference(['B8', 'B12'])
ndwi = sentinel.normalizedDifference(['B3', 'B11'])
```

```python
index_list = [ndvi, nbr, ndwi]
index_names = ['NDVI', 'NBR', 'NDWI'] # Define names for each index

for i, index in enumerate(index_list):
  Map.addLayer(index, {}, index_names[i]) # Use the index name when adding the layer
```

```python
img = ee.Image(sentinel)
new_bands = ee.Image([elevation, ndvi, nbr, ndwi, brightnessTemperature, maximumTemperature, minimumTemperature, precipitationAmount, windSpeed, specificHumidity]).float()
print(type(new_bands))
img = img.addBands(new_bands)
print(type(img))
```

```python
# A Landsat 8 surface reflectance image.
image = img

# A region of interest.
region = aoi

# Set the export "scale" and "crs" parameters.
task = ee.batch.Export.image.toDrive(
    image=image,
    description='image_export',
    folder='ee_demos',
    region=region,
    maxPixels=1e13  #Reduzir
)
task.start()
```
